```python
def calculate_summary_statistics(data):
    if not data:
        return None, None, None
    mean_val = sum(data) / len(data)
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 1:
        median_val = sorted_data[n // 2]
    else:
        median_val = (sorted_data[(n // 2) - 1] + sorted_data[n // 2]) / 2

    frequency = {}
    for num in data:
        frequency[num] = frequency.get(num, 0) + 1
    max_freq = max(frequency.values())

    if max_freq == 1 and len(data) > 1:
        mode_val = "No unique mode (all values appear once)"
    else:
        mode_val = "only single element is present in the input"
    difference=0
    for i in range(len(data)):
        c=data[i]-mean_val
        difference+=c**2
    varience=difference/len(data)
    standard_deviation=varience**0.5

    return mean_val, median_val, mode_val,varience,standard_deviation

x1 = [2, 4, 5, 6, 7, 8, 9, 10]
mean, median, mode ,varience ,standard_deviation = calculate_summary_statistics(x1)
print(f"Dataset: {x1}")
print(f"Mean:   {mean}")
print(f"Median: {median}")
print(f"Mode:   {mode}")
print(f"varience:   {varience}")
print(f"standard_deviation:   {standard_deviation}")
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

median_volume = df['volume'].median()
df['volume_category'] = df['volume'].apply(lambda x: 'High_Volume' if x > median_volume else 'Low_Volume')

df['daily_close_change'] = df['close'].diff()
df['price_change_category'] = df['daily_close_change'].apply(lambda x: 'Increase' if x > 0 else ('Decrease' if x < 0 else 'No_Change'))

contingency_table = pd.crosstab(df['volume_category'], df['price_change_category'], dropna=True)

if contingency_table.shape[0] > 1 and contingency_table.shape[1] > 1:
    print("Contingency Table:")
    display(contingency_table)

    chi2, p, dof, expected = chi2_contingency(contingency_table)

    print(f"\nChi-square Statistic: {chi2:.4f}")
    print(f"P-value: {p:.4f}")
    print(f"Degrees of Freedom: {dof}")

    alpha = 0.05
    if p < alpha:
        print("There is a statistically significant association between volume category and daily close price change category.")
    else:
        print("There is no statistically significant association between volume category and daily close price change category.")
else:
    print("Not enough variation in categories to perform Chi-square test.")
```

### Chi-square Test with a Small Custom Dataset

```python
from scipy.stats import chi2_contingency
import pandas as pd

data_tiny = {
    'Color_Preference': ['Red', 'Blue', 'Red', 'Blue', 'Red', 'Blue'],
    'Gender': ['Male', 'Male', 'Female', 'Female', 'Male', 'Female']
}
df_tiny = pd.DataFrame(data_tiny)

print("Tiny Custom Dataset:")
display(df_tiny)

contingency_table_tiny = pd.crosstab(df_tiny['Gender'], df_tiny['Color_Preference'])
print("\nTiny Custom Contingency Table:")
display(contingency_table_tiny)

chi2_tiny, p_tiny, dof_tiny, expected_tiny = chi2_contingency(contingency_table_tiny)

print(f"\nChi-square Statistic (Tiny): {chi2_tiny:.4f}")
print(f"P-value (Tiny): {p_tiny:.4f}")
print(f"Degrees of Freedom (Tiny): {dof_tiny}")
print("Expected Frequencies (Tiny):")
display(pd.DataFrame(expected_tiny, index=contingency_table_tiny.index, columns=contingency_table_tiny.columns))

alpha_tiny = 0.05
if p_tiny < alpha_tiny:
    print("\nConclusion (Tiny): Reject the null hypothesis. There is a statistically significant association between Gender and Color Preference.")
else:
    print("\nConclusion (Tiny): Fail to reject the null hypothesis. There is no statistically significant association between Gender and Color Preference.")
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

transactions = [
    ['milk', 'bread', 'butter'],
    ['milk', 'water', 'sugar'],
    ['bread', 'water', 'yogurt'],
    ['milk', 'bread', 'water', 'cheese'],
    ['bread', 'butter', 'sugar'],
    ['milk', 'bread', 'butter', 'water']
]

ohe_df = pd.DataFrame(columns=sorted(list(set(item for sublist in transactions for item in sublist))))

for i, transaction in enumerate(transactions):
    row = {item: True for item in transaction}
    ohe_df = pd.concat([ohe_df, pd.DataFrame([row], columns=ohe_df.columns)], ignore_index=True)

ohe_df = ohe_df.fillna(False)

print("One-hot encoded DataFrame (first 5 rows):\n", ohe_df.head())

frequent_itemsets = apriori(ohe_df, min_support=0.5, use_colnames=True)

print("\nFrequent Itemsets (min_support=0.5):\n", frequent_itemsets)

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print("\nAssociation Rules (min_confidence=0.7):\n", rules)
```

```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.frequent_patterns import association_rules

transactions = [
    ['milk', 'bread', 'butter'],
    ['milk', 'water', 'sugar'],
    ['bread', 'water', 'yogurt'],
    ['milk', 'bread', 'water', 'cheese'],
    ['bread', 'butter', 'sugar'],
    ['milk', 'bread', 'butter', 'water']
]

ohe_df = pd.DataFrame(columns=sorted(list(set(item for sublist in transactions for item in sublist))))

for i, transaction in enumerate(transactions):
    row = {item: True for item in transaction}
    ohe_df = pd.concat([ohe_df, pd.DataFrame([row], columns=ohe_df.columns)], ignore_index=True)

ohe_df = ohe_df.fillna(False).infer_objects(copy=False)

print("One-hot encoded DataFrame (first 5 rows):\n", ohe_df.head())

frequent_itemsets_fpgrowth = fpgrowth(ohe_df, min_support=0.5, use_colnames=True)

print("\nFrequent Itemsets (FP-Growth, min_support=0.5):\n", frequent_itemsets_fpgrowth)

rules_fpgrowth = association_rules(frequent_itemsets_fpgrowth, metric="confidence", min_threshold=0.7)

print("\nAssociation Rules (FP-Growth, min_confidence=0.7):\n", rules_fpgrowth)
```

# **Decision Tree**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
df_weather = pd.read_csv('/content/weatherHistory.csv')
```

```python
print("Columns in dataset:", df_weather.columns.tolist())
```

```python
target_col = 'Precip Type'
feature_cols = ['Temperature (C)', 'Apparent Temperature (C)', 'Humidity', 'Wind Speed (km/h)', 'Pressure (millibars)']
```

```python
df_cleaned = df_weather.dropna(subset=feature_cols + [target_col])
```

```python
X = df_cleaned[feature_cols]
y = df_cleaned[target_col].astype(str)
```

```python
class_names = sorted(y.unique())
print(f"Unique target classes to predict: {class_names}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
clf = DecisionTreeClassifier(max_depth=4, random_state=42)
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
plt.figure(figsize=(7, 5))
plot_tree(clf, feature_names=feature_cols, class_names=class_names, filled=True, rounded=True)
plt.title("Decision Tree Induction on Weather Dataset (Precip Type Classification)")
plt.show()
```

### 2. K-Nearest Neighbors (KNN) Classification (Self-Contained)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

df_weather = pd.read_csv('/content/weatherHistory.csv')

target_col = 'Precip Type'
feature_cols = ['Temperature (C)', 'Apparent Temperature (C)', 'Humidity', 'Wind Speed (km/h)', 'Pressure (millibars)']

df_cleaned = df_weather.dropna(subset=feature_cols + [target_col])

X = df_cleaned[feature_cols]
y = df_cleaned[target_col].astype(str)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn_clf = KNeighborsClassifier(n_neighbors=5)
knn_clf.fit(X_train_scaled, y_train)

y_pred_knn = knn_clf.predict(X_test_scaled)
print("=== KNN Performance ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_knn):.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred_knn))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn))
```

### K-Means and K-Medoids Clustering on Weather Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn_extra.cluster import KMedoids
```

```python
df_weather = pd.read_csv('/content/weatherHistory.csv')
feature_cols = ['Temperature (C)', 'Humidity', 'Wind Speed (km/h)', 'Pressure (millibars)']
df_cleaned = df_weather.dropna(subset=feature_cols)
```

```python
df_sample = df_cleaned.sample(n=2000, random_state=42).copy()
X_cluster = df_sample[feature_cols]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_sample['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
kmedoids = KMedoids(n_clusters=3, random_state=42)
df_sample['KMedoids_Cluster'] = kmedoids.fit_predict(X_scaled)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.scatterplot(
    ax=axes[0],
    data=df_sample,
    x='Temperature (C)',
    y='Humidity',
    hue='KMeans_Cluster',
    palette='viridis',
    alpha=0.7
)
axes[0].set_title('K-Means Clustering (k=3)')

sns.scatterplot(
    ax=axes[1],
    data=df_sample,
    x='Temperature (C)',
    y='Humidity',
    hue='KMedoids_Cluster',
    palette='plasma',
    alpha=0.7
)
axes[1].set_title('K-Medoids Clustering (k=3)')

plt.tight_layout()
plt.show()
```
